```python
import sys  
sys.path.insert(1, 'D:/KINEPY/new 22.02.2025/kinepy-dev/kinepy-dev')
import kinepy as kp
import numpy as np
import matplotlib.pyplot as plt
```

```python
L = 7.55
```

```python
bridge = kp.System()

s0 = bridge.ground
s1 = bridge.add_solid()
s2 = bridge.add_solid(mass=1000, g=(L/2, 0))
s3 = bridge.add_solid()

r1 = bridge.add_revolute(s0, s1)
r2 = bridge.add_revolute(s1, s2, p1=(L, 0))
r3 = bridge.add_revolute(s1, s3, p1=(1.5*L, np.sqrt(3)/2*L))
r4 = bridge.add_revolute(s2, s3, p1=(L, 0), p2=(L, 0))
ps = bridge.add_pin_slot(s2, s0, p2=(3*L, 0))

bridge.add_gravity(g=(0, -10))

bridge.determine_computation_order()
n = 2
bridge.set_sim_parameters(n, 3)
```

```python
a, b = bridge.get_steps_with_multiple_solutions()
a.solution_index = 1
b.solution_index = 1
```

```python
bridge.solve_kinematics()
```

```python
kd = bridge.kinematic_diagram()
kd.show()
```

```python
bridge.solve_dynamics()
```
